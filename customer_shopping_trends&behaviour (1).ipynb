Gender distribution

```python
df["Gender"].value_counts()
```

```python
sns.countplot(x="Gender", data=df,color="brown")
plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Customers")
plt.show()
```

Category distribution

```python
df1=df["Category"].value_counts()
df1
```

```python

plt.pie(df1, labels=df1.index, autopct="%1.1f%%")

plt.title("Purchase Distribution by Category")
plt.show()
```

age distribution

```python
df["Age"].value_counts().sort_index()
```

```python
plt.hist(df["Age"], bins=10,color="black")
plt.title("Distribution of Customer Age")
plt.xlabel("Age")
plt.ylabel("Number of Customers")
plt.show()
```

Age vs Purchase Amount

```python
plt.scatter(df["Age"], df["Purchase Amount (USD)"])

plt.title("Age vs Purchase Amount")
plt.xlabel("Age")
plt.ylabel("Purchase Amount (USD)")
plt.show()
```
